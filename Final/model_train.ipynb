```python
import cv2
import os
import re

def extract_frames(video_path, output_folder, frame_interval=5):
    os.makedirs(output_folder, exist_ok=True)

    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print(f"❌ Cannot open: {video_path}")
        return

    frame_count = 0
    saved_count = 0

    print(f"▶ Processing started: {video_path}")

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        if frame_count % frame_interval == 0:
            filename = os.path.join(
                output_folder,
                f"{os.path.basename(video_path)}_{saved_count:05d}.jpg"
            )
            cv2.imwrite(filename, frame)
            saved_count += 1

        # 🔥 PRINT PROGRESS EVERY 100 FRAMES
        if frame_count % 100 == 0:
            print(f"   ⏳ {frame_count} frames processed...")

        frame_count += 1

    cap.release()
    print(f"✅ DONE: {video_path} → {saved_count} frames\n")


# 🔥 YOUR ACTUAL FOLDER PATH
video_folder = r"D:\FYP_videos\videos"

print("🚀 Starting frame extraction...\n")

files = os.listdir(video_folder)

for file in files:
    if file.lower().endswith(".mp4"):

        print(f"📹 Processing: {file}")

        video_path = os.path.join(video_folder, file)

        # ✅ FIXED REGEX (handles 11, 12, 63 etc)
        match = re.search(r'camera_c(\d+)', file)

        if match:
            camera_id = match.group(1)   # e.g., 11, 12, 63
        else:
            print(f"⚠️ Skipping {file} (invalid name)")
            continue

        output_folder = f"dataset/frames/camera_{camera_id}"

        extract_frames(video_path, output_folder, frame_interval=5)

print("\n✅ All videos processed!")
```
