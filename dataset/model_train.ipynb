```python
import cv2
import os

def extract_frames(video_path, output_folder, frame_interval=5):
    os.makedirs(output_folder, exist_ok=True)

    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print(f"❌ Cannot open video: {video_path}")
        return

    frame_count = 0
    saved_count = 0

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        if frame_count % frame_interval == 0:
            filename = os.path.join(output_folder, f"frame_{saved_count:05d}.jpg")
            cv2.imwrite(filename, frame)
            saved_count += 1

        frame_count += 1

    cap.release()
    print(f"✅ Extracted {saved_count} frames from {video_path}")


# UPDATE PATHS HERE
videos = [
    "camera_c1.mp4",
    "camera_c2.mp4",
    "camera_c3.mp4",
]

for i, video in enumerate(videos):
    extract_frames(video, f"dataset/frames/camera_{i+1}", frame_interval=5)
```
